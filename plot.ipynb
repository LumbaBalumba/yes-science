```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
t, traj = np.load('samples.npy')[3]
```

```python
plt.figure(figsize=(20, 12))
plt.xlabel('Время')
plt.ylabel('Концентрация')
plt.plot(t, traj)
```

```python
from pharmapynetics.models import EPBFTPK
```

```python
model = EPBFTPK(tau_estimation_method='minmax')
```

```python
model.fit(t, traj)
```

```python
t_est = np.linspace(np.min(t), np.max(t), 1000)
```

```python
traj_est = model.sample(t_est)
```

```python
plt.figure(figsize=(20, 12))
plt.xlabel('Время')
plt.ylabel('Концентрация')
plt.plot(t, traj, label='Истинная траектория')
plt.plot(t_est, traj_est, label='Оценка траектории')
plt.legend()
```
